```

using System;
using System.Collections.Generic;
using System.Linq;

// Демонстрация работы программы
Computer basicPC = ComputerFactory.CreateBasicPC();
Computer gamingPC = ComputerFactory.CreateGamingPC();

Console.WriteLine("=== БАЗОВЫЙ КОМПЬЮТЕР ===");
basicPC.ShowComponents();
Console.WriteLine("Общая стоимость: " + basicPC.GetTotalPrice() + " руб.");

Console.WriteLine();

Console.WriteLine("=== ИГРОВОЙ КОМПЬЮТЕР ===");
gamingPC.ShowComponents();
Console.WriteLine("Общая стоимость: " + gamingPC.GetTotalPrice() + " руб.");

Console.WriteLine();

Console.WriteLine("=== ИНФОРМАЦИЯ О КОМПОНЕНТАХ ===");
CPU cpu1 = new CPU("Intel Core i5", 18000, 6);
CPU cpu2 = new CPU("Intel Core i7", 28000, 8);

Console.WriteLine(cpu1.GetInfo());
Console.WriteLine(cpu2.GetInfo());

Console.WriteLine();

Console.WriteLine("=== СРАВНЕНИЕ КОМПОНЕНТОВ ===");
Console.WriteLine("Процессоры равны: " + (cpu1 == cpu2));
Console.WriteLine("Процессоры не равны: " + (cpu1 != cpu2));

Console.WriteLine();

Console.WriteLine("=== ДОБАВЛЕНИЕ КОМПОНЕНТА ===");
basicPC.AddComponent(cpu1);
basicPC.ShowComponents();
Console.WriteLine("Общая стоимость: " + basicPC.GetTotalPrice() + " руб.");

Console.WriteLine();

Console.WriteLine("=== УДАЛЕНИЕ КОМПОНЕНТА ===");
basicPC.RemoveComponent(cpu1);
basicPC.ShowComponents();
Console.WriteLine("Общая стоимость: " + basicPC.GetTotalPrice() + " руб.");

Console.WriteLine();

Console.WriteLine("=== ПОИСК КОМПОНЕНТОВ ПО ТИПУ ===");
gamingPC.ShowComponents<GPU>();

Console.WriteLine();

Console.WriteLine("=== СОЗДАНИЕ КОМПЬЮТЕРА ЧЕРЕЗ ОБОБЩЁННУЮ ФАБРИКУ ===");
Computer cpuComputer = ComputerFactory.CreateComputer(
    new CPU("AMD Ryzen 5", 20000, 6),
    new CPU("Intel Core i5", 18000, 6)
);
cpuComputer.ShowComponents();


// Интерфейс устанавливаемого компонента
public interface IInstallable
{
    void Install();
}


// Абстрактный базовый класс
public abstract class Component : IInstallable
{
    public string Name { get; set; }
    public decimal Price { get; set; }

    public Component(string name, decimal price)
    {
        Name = name;
        Price = price;
    }

    public virtual string GetInfo()
    {
        return "Название: " + Name + ", цена: " + Price + " руб.";
    }

    public override string ToString()
    {
        return GetInfo();
    }

    public abstract void Install();
}


// Процессор
public class CPU : Component
{
    public int Cores { get; set; }

    public CPU(string name, decimal price, int cores)
        : base(name, price)
    {
        Cores = cores;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + ", количество ядер: " + Cores;
    }

    public override void Install()
    {
        Console.WriteLine("Установка процессора: " + Name);
    }

    public static bool operator ==(CPU a, CPU b)
    {
        if (ReferenceEquals(a, b))
            return true;

        if (a is null || b is null)
            return false;

        return a.Name == b.Name && a.Price == b.Price && a.Cores == b.Cores;
    }

    public static bool operator !=(CPU a, CPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is CPU other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price, Cores);
    }
}


// Оперативная память
public class RAM : Component
{
    public int Capacity { get; set; }

    public RAM(string name, decimal price, int capacity)
        : base(name, price)
    {
        Capacity = capacity;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + ", объём памяти: " + Capacity + " ГБ";
    }

    public override void Install()
    {
        Console.WriteLine("Установка оперативной памяти: " + Name);
    }

    public static bool operator ==(RAM a, RAM b)
    {
        if (ReferenceEquals(a, b))
            return true;

        if (a is null || b is null)
            return false;

        return a.Name == b.Name && a.Price == b.Price && a.Capacity == b.Capacity;
    }

    public static bool operator !=(RAM a, RAM b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is RAM other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price, Capacity);
    }
}


// Видеокарта
public class GPU : Component
{
    public int Memory { get; set; }

    public GPU(string name, decimal price, int memory)
        : base(name, price)
    {
        Memory = memory;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + ", видеопамять: " + Memory + " ГБ";
    }

    public override void Install()
    {
        Console.WriteLine("Установка видеокарты: " + Name);
    }

    public static bool operator ==(GPU a, GPU b)
    {
        if (ReferenceEquals(a, b))
            return true;

        if (a is null || b is null)
            return false;

        return a.Name == b.Name && a.Price == b.Price && a.Memory == b.Memory;
    }

    public static bool operator !=(GPU a, GPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is GPU other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price, Memory);
    }
}


// Жёсткий диск
public class HDD : Component
{
    public int Capacity { get; set; }

    public HDD(string name, decimal price, int capacity)
        : base(name, price)
    {
        Capacity = capacity;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + ", объём диска: " + Capacity + " ГБ";
    }

    public override void Install()
    {
        Console.WriteLine("Установка жёсткого диска: " + Name);
    }

    public static bool operator ==(HDD a, HDD b)
    {
        if (ReferenceEquals(a, b))
            return true;

        if (a is null || b is null)
            return false;

        return a.Name == b.Name && a.Price == b.Price && a.Capacity == b.Capacity;
    }

    public static bool operator !=(HDD a, HDD b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is HDD other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price, Capacity);
    }
}


// Класс компьютера
public class Computer
{
    private List<Component> components = new List<Component>();

    public void AddComponent<T>(T component)
        where T : Component, IInstallable
    {
        if (component == null)
        {
            Console.WriteLine("Нельзя добавить пустой компонент.");
            return;
        }

        component.Install();
        components.Add(component);
        Console.WriteLine("Компонент добавлен: " + component.Name);
    }

    public void AddComponentFromFactory(Component component)
    {
        if (component == null)
            return;

        component.Install();
        components.Add(component);
    }

    public void RemoveComponent<T>(T component)
        where T : Component
    {
        if (components.Remove(component))
            Console.WriteLine("Компонент удалён: " + component.Name);
        else
            Console.WriteLine("Компонент не найден.");
    }

    public decimal GetTotalPrice()
    {
        return components.Sum(component => component.Price);
    }

    public void ShowComponents()
    {
        if (components.Count == 0)
        {
            Console.WriteLine("Компьютер не содержит компонентов.");
            return;
        }

        foreach (Component component in components)
        {
            Console.WriteLine(component.GetInfo());
        }
    }

    public void ShowComponents<T>() where T : Component
    {
        bool found = false;

        foreach (Component component in components)
        {
            if (component is T)
            {
                Console.WriteLine(component.GetInfo());
                found = true;
            }
        }

        if (!found)
            Console.WriteLine("Компоненты данного типа не найдены.");
    }
}


// Фабрика компьютеров
public static class ComputerFactory
{
    public static Computer CreateComputer<T>(params T[] components)
        where T : Component, IInstallable
    {
        Computer computer = new Computer();

        foreach (T component in components)
        {
            computer.AddComponent(component);
        }

        return computer;
    }

    public static Computer CreateBasicPC()
    {
        Computer computer = new Computer();

        computer.AddComponentFromFactory(new CPU("Intel Core i3", 12000, 4));
        computer.AddComponentFromFactory(new RAM("Kingston", 4000, 8));
        computer.AddComponentFromFactory(new HDD("Seagate", 5000, 1000));

        return computer;
    }

    public static Computer CreateGamingPC()
    {
        Computer computer = new Computer();

        computer.AddComponentFromFactory(new CPU("AMD Ryzen 7", 30000, 8));
        computer.AddComponentFromFactory(new RAM("Kingston Fury", 8000, 32));
        computer.AddComponentFromFactory(new GPU("NVIDIA RTX 4070", 65000, 12));
        computer.AddComponentFromFactory(new HDD("WD Black", 7000, 2000));

        return computer;
    }
}

```
