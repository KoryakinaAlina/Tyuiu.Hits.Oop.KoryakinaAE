```

using System;
using System.Collections.Generic;
using System.Linq;

public class LineItem
{
    public string Name { get; set; }
    public double Price { get; set; }
    public int Quantity { get; set; } = 1;
    public string Category { get; set; } = "Общее";
    public string Description { get; set; } = "";

    public LineItem(string name, double price)
    {
        Name = name;
        Price = price;
    }

    public double GetTotal() => Price * Quantity;
    public void ChangePrice(double price) => Price = price;
    public void ChangeName(string name) => Name = name;
}

public class Invoice
{
    public string InvoiceNumber { get; set; }
    public DateTime IssueDate { get; set; }
    public double TotalAmount { get; set; }
    public List<LineItem> Lines { get; set; } = new List<LineItem>();

    public string Customer { get; set; } = "";
    public string Status { get; set; } = "Создана";
    public string Currency { get; set; } = "руб.";

    public Invoice(string number, DateTime date)
    {
        InvoiceNumber = number;
        IssueDate = date;
    }

    public virtual double CalculateTotal()
    {
        TotalAmount = Lines.Sum(x => x.Price);
        return TotalAmount;
    }

    public virtual void AddLine(LineItem item)
    {
        Lines.Add(item);
        Console.WriteLine($"Добавлена позиция: {item.Name}");
    }

    public virtual void RemoveLine(LineItem item)
    {
        Lines.Remove(item);
        Console.WriteLine($"Удалена позиция: {item.Name}");
    }

    public int GetLineCount() => Lines.Count;
    public void ChangeStatus(string status) => Status = status;
    public LineItem FindLine(string name) => Lines.Find(x => x.Name == name);
}

// Простое наследование
public class GoodsInvoice : Invoice
{
    public DateTime SupplyDate { get; set; }
    public string Supplier { get; set; } = "";
    public string Address { get; set; } = "";
    public bool Delivered { get; set; }

    public GoodsInvoice(string n, DateTime d, DateTime supply)
        : base(n, d)
    {
        SupplyDate = supply;
    }

    public override void AddLine(LineItem item)
    {
        base.AddLine(item);
        Console.WriteLine($"Дата поставки товара: {SupplyDate:dd.MM.yyyy}");
    }

    public void Deliver() => Delivered = true;
    public void ChangeAddress(string address) => Address = address;
    public void ShowSupplyDate() => Console.WriteLine(SupplyDate.ToShortDateString());
}

// Сложное наследование: цепочка классов
public class ElectronicGoodsInvoice : GoodsInvoice
{
    public string Manufacturer { get; set; } = "";
    public int Warranty { get; set; }
    public string SerialNumber { get; set; } = "";

    public ElectronicGoodsInvoice(string n, DateTime d, DateTime supply)
        : base(n, d, supply) { }

    public void ExtendWarranty(int months) => Warranty += months;
    public void ShowManufacturer() => Console.WriteLine(Manufacturer);
    public void ChangeSerial(string number) => SerialNumber = number;
}

public class ServiceInvoice : Invoice
{
    public DateTime ServiceDate { get; set; }
    public string Provider { get; set; } = "";
    public string Reason { get; set; } = "";

    public ServiceInvoice(string n, DateTime d, DateTime service)
        : base(n, d)
    {
        ServiceDate = service;
    }

    public override void RemoveLine(LineItem item)
    {
        base.RemoveLine(item);
        Console.WriteLine($"Причина аннулирования услуги: {Reason}.");
    }

    public void ChangeProvider(string provider) => Provider = provider;
    public void ChangeDate(DateTime date) => ServiceDate = date;
    public void SetReason(string reason) => Reason = reason;
}

// Множественное наследование через интерфейсы
public interface IReturnable
{
    void ProcessReturn(double amount);
}

public interface IPrintable
{
    void PrintInvoice();
}

public class CombinedInvoice : Invoice, IReturnable, IPrintable
{
    public bool ReturnAllowed { get; set; }
    public double ReturnedAmount { get; set; }
    public string PaymentMethod { get; set; } = "Наличные";

    public CombinedInvoice(string n, DateTime d, bool allowed)
        : base(n, d)
    {
        ReturnAllowed = allowed;
    }

    public override double CalculateTotal()
    {
        TotalAmount = Lines.Sum(x => x.Price);
        return TotalAmount;
    }

    public void ProcessReturn(double amount)
    {
        if (ReturnAllowed && amount > 0 && amount <= CalculateTotal())
            ReturnedAmount += amount;
    }

    public void PrintInvoice() =>
        Console.WriteLine($"Фактура №{InvoiceNumber}: {CalculateTotal()} руб.");

    public void ShowItemFromInvoice(LineItem item)
    {
        if (Lines.Contains(item))
            Console.WriteLine($"В фактуре №{InvoiceNumber} находится позиция '{item.Name}' стоимостью {item.Price} руб.");
    }
}

// Основная программа
LineItem product = new LineItem("Ноутбук", 50000);
LineItem service = new LineItem("Настройка программы", 5000);

GoodsInvoice goods = new GoodsInvoice("G-001", DateTime.Today,
    new DateTime(2026, 9, 15));
goods.AddLine(product);
Console.WriteLine($"Общая сумма фактуры: {goods.CalculateTotal()} руб.");
Console.WriteLine();

ServiceInvoice services = new ServiceInvoice("S-001", DateTime.Today, DateTime.Today);
services.AddLine(service);
Console.WriteLine($"Общая сумма фактуры: {services.CalculateTotal()} руб.");
services.SetReason("услуга отменена");
services.RemoveLine(service);
Console.WriteLine();

CombinedInvoice combined = new CombinedInvoice("C-001", DateTime.Today, true);
combined.AddLine(product);
combined.AddLine(service);
Console.WriteLine($"Общая сумма фактуры: {combined.CalculateTotal()} руб.");
Console.WriteLine("Возврат товара или услуги разрешен.");
Console.WriteLine();

combined.ShowItemFromInvoice(product);
combined.ShowItemFromInvoice(service);

// Дополнительные методы и атрибуты
Console.WriteLine();
Console.WriteLine("Дополнительные результаты:");

product.Quantity = 2;
Console.WriteLine($"Стоимость двух ноутбуков: {product.GetTotal()} руб.");
Console.WriteLine($"Количество позиций в фактуре: {combined.GetLineCount()}");

ElectronicGoodsInvoice electronic = new ElectronicGoodsInvoice(
    "E-001", DateTime.Today, DateTime.Today);
electronic.Manufacturer = "HP";
electronic.Warranty = 24;
electronic.ShowManufacturer();
Console.WriteLine($"Гарантия: {electronic.Warranty} месяцев");

services.Provider = "Сервисный центр";
services.ChangeProvider("Новый сервисный центр");
Console.WriteLine($"Исполнитель: {services.Provider}");

combined.ProcessReturn(5000);
combined.PrintInvoice();
```
