```

using System;

// Базовый класс — сотрудник
public class Employee
{
    private string Name { get; set; }
    private string Position { get; set; }
    private decimal Salary { get; set; }

    public Employee(string name, string position, decimal salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    public virtual void PrintInfo()
    {
        Console.WriteLine($"Имя сотрудника: {Name}, Должность: {Position}, Зарплата: {Salary} руб.");
    }
}

// Простое наследование — менеджер
public class Manager : Employee
{
    public string Department { get; set; }

    public Manager(string name, string position, decimal salary, string department)
        : base(name, position, salary)
    {
        Department = department;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Отдел управления: {Department}");
    }
}

// Сложное наследование — проектный менеджер
public class ProjectManager : Manager
{
    public string ProjectName { get; set; }

    public ProjectManager(string name, string position, decimal salary,
        string department, string projectName)
        : base(name, position, salary, department)
    {
        ProjectName = projectName;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Название проекта: {ProjectName}");
    }
}

// Интерфейс для создания отчета
public interface IReportable
{
    string Report { get; set; }
    void GenerateReport();
}

// Интерфейс для списка команды
public interface ITeamLeader
{
    string Team { get; set; }
    void GetTeamMembers();
}

// Класс, реализующий оба интерфейса
public class SuperProjectManager : ProjectManager, IReportable, ITeamLeader
{
    public string Report { get; set; }
    public string Team { get; set; }

    public SuperProjectManager(string name, string position, decimal salary,
        string department, string projectName, string report, string team)
        : base(name, position, salary, department, projectName)
    {
        Report = report;
        Team = team;
    }

    public void GenerateReport()
    {
        Console.WriteLine($"Отчет: {Report}");
    }

    public void GetTeamMembers()
    {
        Console.WriteLine($"Команда: {Team}");
    }
}

// Проверка работы классов

Manager manager = new Manager(
    "Алина", "Менеджер", 100000, "Отдел продаж");

manager.PrintInfo();
Console.WriteLine();

ProjectManager projectManager = new ProjectManager(
    "Кристина", "Проектный менеджер", 150000,
    "Отдел проектирования", "Детский мир");

projectManager.PrintInfo();
Console.WriteLine();

SuperProjectManager superManager = new SuperProjectManager(
    "Валерия", "Главный проектный менеджер", 200000,
    "Отдел планирования", "Детский мир",
    "Задачи на сегодня выполнены",
    "Алина, Кристина, Валерия");

superManager.PrintInfo();
superManager.GenerateReport();
superManager.GetTeamMembers();
```
