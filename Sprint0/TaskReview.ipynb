```
 using System;

public class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; }

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Bloom()
    {
        Console.WriteLine($"Растение {Name} цветёт.");
    }

    public void Grow(float growth)
    {
        Height += growth;
        Console.WriteLine($"Высота растения увеличилась на {growth} см. Теперь: {Height} см.");
    }
}

public class Flower : Plant
{
    public string Color { get; set; }
    public int PetalCount { get; set; }

    public Flower(string name, float height, string color, int petalCount)
        : base(name, "Цветок", height)
    {
        Color = color;
        PetalCount = petalCount;
    }

    public override void Bloom()
    {
        Console.WriteLine($"На растении {Name} распустились цветы!");
    }

    public void Wilt()
    {
        Console.WriteLine($"Цветок {Name} постепенно увядает.");
    }
}

public class Tree : Plant
{
    public int Age { get; set; }
    public string LeafType { get; set; }

    public Tree(string name, float height, int age, string leafType)
        : base(name, "Дерево", height)
    {
        Age = age;
        LeafType = leafType;
    }

    public override void Bloom()
    {
        Console.WriteLine($"У дерева {Name} появились новые листья!");
    }

    public void GrowYears(int years)
    {
        Age += years;
        Console.WriteLine($"Возраст дерева увеличился. Сейчас {Age} лет.");
    }
}

// Создание объектов
Flower flower = new Flower("Лилия", 35, "Белый", 6);
Tree tree = new Tree("Дуб", 300, 20, "Крупные листья");

// Информация о цветке
Console.WriteLine("=== Информация о цветке ===");
Console.WriteLine($"Название: {flower.Name}");
Console.WriteLine($"Цвет: {flower.Color}");
Console.WriteLine($"Лепестков: {flower.PetalCount}");
Console.WriteLine($"Начальная высота: {flower.Height} см");

Console.WriteLine("\nДействия с цветком:");
flower.Bloom();
flower.Grow(15);
flower.Wilt();

// Информация о дереве
Console.WriteLine("\n=== Информация о дереве ===");
Console.WriteLine($"Название: {tree.Name}");
Console.WriteLine($"Тип: {tree.Type}");
Console.WriteLine($"Возраст: {tree.Age} лет");
Console.WriteLine($"Вид листьев: {tree.LeafType}");
Console.WriteLine($"Начальная высота: {tree.Height} см");

Console.WriteLine("\nДействия с деревом:");
tree.Bloom();
tree.Grow(40);
tree.GrowYears(5);
```
