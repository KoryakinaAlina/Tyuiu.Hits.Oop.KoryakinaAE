```

using System;
using System.Collections.Generic;

public class Product
{
    private string name;
    private decimal price;
    private int quantity;

    protected string category;
    internal string description;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public decimal Price
    {
        get { return price; }
        set { price = value; }
    }

    public int Quantity
    {
        get { return quantity; }
        set { quantity = value; }
    }

    public Product(
        string name,
        decimal price,
        int quantity,
        string category,
        string description)
    {
        this.name = name;
        this.price = price;
        this.quantity = quantity;
        this.category = category;
        this.description = description;
    }

    public decimal GetTotalPrice()
    {
        return price * quantity;
    }

    public void UpdateQuantity(int amount)
    {
        quantity += amount;
    }

    public string GetProductInfo()
    {
        return $"Товар: {name}, цена: {price} руб., " +
               $"количество: {quantity}, категория: {category}, " +
               $"описание: {description}";
    }
}


public class Seller
{
    private string name;
    private string employeeId;

    protected decimal salary;
    internal string contactInfo;

    private List<Product> products;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public string EmployeeId
    {
        get { return employeeId; }
        set { employeeId = value; }
    }

    public Seller(
        string name,
        string employeeId,
        decimal salary,
        string contactInfo)
    {
        this.name = name;
        this.employeeId = employeeId;
        this.salary = salary;
        this.contactInfo = contactInfo;

        products = new List<Product>();
    }

    public void AddProduct(Product product)
    {
        products.Add(product);

        Console.WriteLine(
            $"Товар '{product.Name}' добавлен продавцом {name}.");
    }

    public void SellProduct(Product product, int quantity)
    {
        product.UpdateQuantity(-quantity);

        Console.WriteLine(
            $"Продавец {name} продал {quantity} шт. товара '{product.Name}'.");
    }

    public string GetSellerInfo()
    {
        return $"Продавец: {name}, ID сотрудника: {employeeId}, " +
               $"зарплата: {salary} руб., контакты: {contactInfo}";
    }
}


public class Store
{
    private string storeName;
    private string location;

    public string storeHours { get; set; }

    private List<Seller> sellers;

    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    public string Location
    {
        get { return location; }
        set { location = value; }
    }

    public Store(
        string storeName,
        string location,
        string storeHours)
    {
        this.storeName = storeName;
        this.location = location;
        this.storeHours = storeHours;

        sellers = new List<Seller>();
    }

    public void AddSeller(Seller seller)
    {
        sellers.Add(seller);

        Console.WriteLine(
            $"Продавец '{seller.Name}' добавлен в магазин '{storeName}'.");
    }

    public void ListProducts()
    {
        Console.WriteLine($"Товары продавцов магазина '{storeName}':");

        foreach (Seller seller in sellers)
        {
            Console.WriteLine(seller.GetSellerInfo());
        }
    }

    public string GetStoreInfo()
    {
        return $"Магазин: {storeName}, адрес: {location}, " +
               $"часы работы: {storeHours}";
    }
}


// Создание объекта товара
Product product = new Product(
    "Ноутбук",
    75000,
    5,
    "Электроника",
    "Ноутбук для работы и учебы");

Console.WriteLine(product.GetProductInfo());
Console.WriteLine($"Общая стоимость товара: {product.GetTotalPrice()} руб.");

Console.WriteLine();


// Создание объекта продавца
Seller seller = new Seller(
    "Корякина Алина",
    "S-001",
    60000,
    "koryakina@mail.ru");

seller.AddProduct(product);
seller.SellProduct(product, 1);

Console.WriteLine(seller.GetSellerInfo());

Console.WriteLine();


// Создание объекта магазина
Store store = new Store(
    "ТИУ",
    "г. Тюмень",
    "08:00 - 21:00");

store.AddSeller(seller);

Console.WriteLine(store.GetStoreInfo());

Console.WriteLine();

store.ListProducts();

Console.WriteLine();
Console.WriteLine(product.GetProductInfo());

```
