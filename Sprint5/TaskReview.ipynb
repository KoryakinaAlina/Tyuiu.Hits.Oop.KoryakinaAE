```
using System;
using System.Collections.Generic;
using System.Linq;

Console.OutputEncoding = System.Text.Encoding.UTF8;

Console.WriteLine("===============================================================================");
Console.WriteLine("=== СПРИНТ РЕВЬЮ 5: СИСТЕМА УЧЕТА ЗАКАЗОВ В КОМПАНИИ ===");
Console.WriteLine("===============================================================================\n");

// ============================================================================
// 1. ИНИЦИАЛИЗАЦИЯ СУЩНОСТЕЙ (КОМПАНИЯ, СОТРУДНИКИ, ЗАКАЗЧИКИ)
// ============================================================================

// Создаем компанию
Company company = new Company("ООО 'Компьютерные Системы и Сервис'");

// Создаем сотрудников
Employee emp1 = new Employee("Иван", "Иванов", "Старший специалист по сборке", 65000);
Employee emp2 = new Employee("Анна", "Петрова", "Ведущий менеджер проектов", 80000);
Employee emp3 = new Employee("Сергей", "Смирнов", "Инженер технической поддержки", 55000);

company.AddEmployee(emp1);
company.AddEmployee(emp2);
company.AddEmployee(emp3);

// Создаем заказчиков
Customer cust1 = new Customer("Алексей", "Кузнецов", "+7 (912) 111-22-33", "ул. Республики, д. 15");
Customer cust2 = new Customer("Елена", "Морозова", "+7 (922) 444-55-66", "ул. Мельникайте, д. 45");
Customer cust3 = new Customer("Дмитрий", "Соколов", "+7 (904) 777-88-99", "ул. Ленина, д. 102");

company.AddCustomer(cust1);
company.AddCustomer(cust2);
company.AddCustomer(cust3);

Console.WriteLine();

// ============================================================================
// 2. СОЗДАНИЕ И НАЗНАЧЕНИЕ ЗАКАЗОВ
// ============================================================================

Console.WriteLine(">>> 1. СОЗДАНИЕ ЗАКАЗОВ И НАЗНАЧЕНИЕ СОТРУДНИКОВ:\n");

// Создание заказов (OrderId генерируется автоматически)
Order order1 = company.CreateOrder(cust1, "Сборка игрового ПК (RTX 4070, Ryzen 7)", 115000);
Order order2 = company.CreateOrder(cust2, "Модернизация офисного ПК (16 ГБ RAM + SSD)", 15000);
Order order3 = company.CreateOrder(cust1, "Установка операционной системы и настройка ПО", 5000);
Order order4 = company.CreateOrder(cust3, "Диагностика и замена видеокарты", 38000);

// Назначение сотрудников на заказы
company.AssignEmployeeToOrder(order1.OrderId, emp1);
company.AssignEmployeeToOrder(order2.OrderId, emp1);
company.AssignEmployeeToOrder(order3.OrderId, emp2);
company.AssignEmployeeToOrder(order4.OrderId, emp3);

Console.WriteLine();

// ============================================================================
// 3. ДЕЛЕГАТЫ, СОБЫТИЯ И ИЗМЕНЕНИЕ СТАТУСОВ ЗАКАЗОВ (Task 2)
// ============================================================================

Console.WriteLine(">>> 2. ИЗМЕНЕНИЕ СТАТУСОВ ЗАКАЗОВ И ОБРАБОТКА СОБЫТИЙ С ДЕЛЕГАТАМИ:\n");

// Перевод в работу (вызов события OnStatusChanged)
company.ChangeOrderStatus(order1.OrderId, OrderStatus.InProgress);
company.ChangeOrderStatus(order2.OrderId, OrderStatus.InProgress);

// Завершение заказов (автоматически увеличивается счетчик OrdersProcessed у сотрудника)
company.ChangeOrderStatus(order1.OrderId, OrderStatus.Completed);
company.ChangeOrderStatus(order2.OrderId, OrderStatus.Completed);

// Отмена заказа
company.ChangeOrderStatus(order3.OrderId, OrderStatus.Cancelled);

Console.WriteLine();

// ============================================================================
// 4. КОЛЛЕКЦИИ И ЛЯМБДА-ВЫРАЖЕНИЯ (Task 0, Task 1, Task 2)
// ============================================================================

Console.WriteLine(">>> 3. ФИЛЬТРАЦИЯ ЗАКАЗОВ ПО СТАТУСУ (ЛЯМБДА-ВЫРАЖЕНИЯ LINQ):");
var completedOrders = company.GetOrdersByStatus(OrderStatus.Completed);
Console.WriteLine($"Завершенные заказы (всего {completedOrders.Count}):");
foreach (var ord in completedOrders)
{
    Console.WriteLine($"   * {ord.GetShortInfo()}");
}
Console.WriteLine();

Console.WriteLine(">>> 4. ПОИСК ЗАКАЗОВ ПО ЗАКАЗЧИКУ:");
var kuznetsovOrders = company.FindOrdersByCustomer("Кузнецов");
Console.WriteLine($"Заказы клиента Кузнецов (всего {kuznetsovOrders.Count}):");
foreach (var ord in kuznetsovOrders)
{
    Console.WriteLine($"   * {ord.GetShortInfo()}");
}
Console.WriteLine();

Console.WriteLine(">>> 5. БЫСТРЫЙ ПОИСК В СЛОВАРЕ DICTIONARY<int, Order> (Task 0):");
if (company.FindOrderById(order1.OrderId, out Order? foundOrder))
{
    Console.WriteLine("Найден заказ в словаре компании:");
    Console.WriteLine($"   {foundOrder!.GetDetailedInfo()}");
}
Console.WriteLine();

Console.WriteLine(">>> 6. КОЛЛЕКЦИЯ УНИКАЛЬНЫХ ЗАКАЗЧИКОВ HASHSET<Customer> (Task 1):");
Customer duplicateCustomer = new Customer("Алексей", "Кузнецов", "+7 (912) 111-22-33");
bool addedDuplicate = company.AddCustomer(duplicateCustomer);
Console.WriteLine($"Попытка добавить дубликат заказчика: {addedDuplicate} (False — отсеян HashSet)");
Console.WriteLine($"Всего уникальных заказчиков в базе: {company.GetUniqueCustomerCount()}\n");

// ============================================================================
// 5. ОТЧЕТЫ ПО СОТРУДНИКАМ
// ============================================================================

Console.WriteLine(">>> 7. ОТЧЕТ ПО ОБРАБОТКЕ ЗАКАЗОВ СОТРУДНИКАМИ (OrdersProcessed):\n");
company.GenerateEmployeeReport();

Console.WriteLine("\n>>> 8. ОБЩИЙ СВОДНЫЙ ОТЧЕТ КОМПАНИИ:\n");
company.GenerateCompanySummary();


// ============================================================================
// ПЕРЕЧИСЛЕНИЕ СТАТУСОВ ЗАКАЗА
// ============================================================================
public enum OrderStatus
{
    Created,        // Создан
    InProgress,     // В работе
    Completed,      // Завершен
    Cancelled       // Отменен
}


// ============================================================================
// ДЕЛЕГАТЫ ДЛЯ УВЕДОМЛЕНИЙ И СОБЫТИЙ (Task 2)
// ============================================================================

// Делегат для уведомления об изменении статуса заказа (событие)
public delegate void OrderStatusChangedHandler(Order order, OrderStatus oldStatus, OrderStatus newStatus);

// Делегат для отправки текстовых уведомлений
public delegate void NotificationSender(string recipient, string message);


// ============================================================================
// БАЗОВЫЙ АБСТРАКТНЫЙ КЛАСС PERSON (ООП: Инкапсуляция и наследование)
// ============================================================================
public abstract class Person
{
    public string FirstName { get; set; }
    public string LastName { get; set; }
    public string PhoneNumber { get; set; }

    public Person(string firstName, string lastName, string phoneNumber)
    {
        FirstName = firstName;
        LastName = lastName;
        PhoneNumber = phoneNumber;
    }

    public string GetFullName() => $"{FirstName} {LastName}";

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"ФИО: {GetFullName()}, Телефон: {PhoneNumber}");
    }

    // Переопределение Equals и GetHashCode для поддержки HashSet (Task 1)
    public override bool Equals(object? obj)
    {
        if (obj is Person other)
        {
            return FirstName == other.FirstName &&
                   LastName == other.LastName &&
                   PhoneNumber == other.PhoneNumber;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(FirstName, LastName, PhoneNumber);
    }
}


// ============================================================================
// КЛАСС СОТРУДНИКА (Employee)
// ============================================================================
public class Employee : Person
{
    public string Position { get; set; }
    public int OrdersProcessed { get; private set; }
    public decimal Salary { get; set; }
    public DateTime HireDate { get; set; }

    public Employee(string firstName, string lastName, string position, decimal salary)
        : base(firstName, lastName, "Внутренний")
    {
        Position = position;
        Salary = salary;
        OrdersProcessed = 0;
        HireDate = DateTime.Today;
    }

    // Метод 1: Увеличение счетчика обработанных заказов
    public void IncrementProcessedOrders()
    {
        OrdersProcessed++;
    }

    // Метод 2: Получение уведомления через делегат
    public void Notify(string message)
    {
        Console.WriteLine($"   [Уведомление сотруднику {GetFullName()}]: {message}");
    }

    // Метод 3: Изменение должности и зарплаты
    public void Promote(string newPosition, decimal newSalary)
    {
        Position = newPosition;
        Salary = newSalary;
        Console.WriteLine($"Сотрудник {GetFullName()} переведен на должность '{Position}' с окладом {Salary} руб.");
    }

    // Метод 4: Вывод информации о сотруднике
    public override void DisplayInfo()
    {
        Console.WriteLine($"Сотрудник: {GetFullName()} | Должность: {Position} | Оклад: {Salary} руб. | Обработано заказов: {OrdersProcessed}");
    }
}


// ============================================================================
// КЛАСС ЗАКАЗЧИКА (Customer)
// ============================================================================
public class Customer : Person
{
    public string Address { get; set; }
    public decimal TotalSpent { get; private set; }
    public int OrdersCount { get; private set; }

    public Customer(string firstName, string lastName, string phoneNumber, string address = "Не указан")
        : base(firstName, lastName, phoneNumber)
    {
        Address = address;
        TotalSpent = 0;
        OrdersCount = 0;
    }

    // Метод 1: Добавление суммы покупки
    public void RegisterPurchase(decimal amount)
    {
        TotalSpent += amount;
        OrdersCount++;
    }

    // Метод 2: Обновление номера телефона
    public void UpdatePhoneNumber(string newPhoneNumber)
    {
        PhoneNumber = newPhoneNumber;
        Console.WriteLine($"Телефон клиента {GetFullName()} обновлен на: {PhoneNumber}");
    }

    // Метод 3: Получение уведомления
    public void Notify(string message)
    {
        Console.WriteLine($"   [SMS для клиента {GetFullName()} ({PhoneNumber})]: {message}");
    }

    // Метод 4: Вывод информации
    public override void DisplayInfo()
    {
        Console.WriteLine($"Заказчик: {GetFullName()} | Тел: {PhoneNumber} | Адрес: {Address} | Заказов: {OrdersCount} | Сумма покупок: {TotalSpent} руб.");
    }
}


// ============================================================================
// КЛАСС ЗАКАЗА (Order)
// ============================================================================
public class Order
{
    private static int _globalIdCounter = 1000;

    public int OrderId { get; private set; }
    public string Description { get; set; }
    public Customer OrderCustomer { get; set; }
    public Employee? AssignedEmployee { get; set; }
    public OrderStatus Status { get; private set; }
    public DateTime CreatedAt { get; private set; }
    public DateTime? CompletedAt { get; private set; }
    public decimal Price { get; set; }

    // Событие изменения статуса заказа (Task 2)
    public event OrderStatusChangedHandler? OnStatusChanged;

    public Order(Customer customer, string description, decimal price)
    {
        OrderId = ++_globalIdCounter;
        OrderCustomer = customer;
        Description = description;
        Price = price;
        Status = OrderStatus.Created;
        CreatedAt = DateTime.Now;
        CompletedAt = null;
        AssignedEmployee = null;
    }

    // Метод 1: Назначение ответственного сотрудника
    public void AssignEmployee(Employee employee)
    {
        AssignedEmployee = employee;
        Console.WriteLine($"На заказ №{OrderId} назначен ответственный: {employee.GetFullName()} ({employee.Position})");
    }

    // Метод 2: Изменение статуса заказа с вызовом события
    public void SetStatus(OrderStatus newStatus)
    {
        if (Status == newStatus) return;

        OrderStatus oldStatus = Status;
        Status = newStatus;

        if (newStatus == OrderStatus.Completed)
        {
            CompletedAt = DateTime.Now;
            AssignedEmployee?.IncrementProcessedOrders();
            OrderCustomer.RegisterPurchase(Price);
        }

        // Генерация события
        OnStatusChanged?.Invoke(this, oldStatus, newStatus);
    }

    // Метод 3: Краткая информация
    public string GetShortInfo()
    {
        string emp = AssignedEmployee != null ? AssignedEmployee.GetFullName() : "Не назначен";
        return $"Заказ №{OrderId} [{Status}] | {Description} | Клиент: {OrderCustomer.GetFullName()} | Ответственный: {emp} | Сумма: {Price} руб.";
    }

    // Метод 4: Детальная информация
    public string GetDetailedInfo()
    {
        string emp = AssignedEmployee != null ? AssignedEmployee.GetFullName() : "Не назначен";
        string completed = CompletedAt.HasValue ? CompletedAt.Value.ToString("dd.MM.yyyy HH:mm") : "В процессе";
        return $"=== ЗАКАЗ №{OrderId} ===\n" +
               $"Описание: {Description}\n" +
               $"Клиент: {OrderCustomer.GetFullName()} (Тел: {OrderCustomer.PhoneNumber})\n" +
               $"Ответственный сотрудник: {emp}\n" +
               $"Статус: {Status}\n" +
               $"Дата создания: {CreatedAt:dd.MM.yyyy HH:mm}\n" +
               $"Дата завершения: {completed}\n" +
               $"Стоимость: {Price:F2} руб.";
    }
}


// ============================================================================
// КЛАСС КОМПАНИИ (Company)
// ============================================================================
public class Company
{
    public string Name { get; set; }

    // Коллекции (List, Dictionary, HashSet по теории Task 0, 1)
    public List<Employee> Employees { get; private set; }
    public List<Order> Orders { get; private set; }
    public List<Customer> Customers { get; private set; }
    
    // Словарь заказов по идентификатору (Task 0)
    private readonly Dictionary<int, Order> _ordersById;

    // Множество уникальных заказчиков (Task 1)
    private readonly HashSet<Customer> _uniqueCustomers;

    // Делегат для уведомлений
    public NotificationSender NotificationService { get; set; }

    public Company(string name)
    {
        Name = name;
        Employees = new List<Employee>();
        Orders = new List<Order>();
        Customers = new List<Customer>();
        _ordersById = new Dictionary<int, Order>();
        _uniqueCustomers = new HashSet<Customer>();

        // Настройка делегата уведомлений
        NotificationService = (recipient, message) =>
        {
            Console.WriteLine($"   [Система Уведомлений компании '{Name}'] -> {recipient}: {message}");
        };
    }

    // 1. Добавление сотрудника
    public void AddEmployee(Employee employee)
    {
        if (employee != null && !Employees.Contains(employee))
        {
            Employees.Add(employee);
            Console.WriteLine($"В компанию '{Name}' добавлен сотрудник: {employee.GetFullName()} ({employee.Position})");
        }
    }

    // 2. Добавление заказчика с контролем уникальности через HashSet
    public bool AddCustomer(Customer customer)
    {
        if (customer == null) return false;

        bool isUnique = _uniqueCustomers.Add(customer);
        if (isUnique)
        {
            Customers.Add(customer);
            Console.WriteLine($"Зарегистрирован новый заказчик: {customer.GetFullName()} ({customer.PhoneNumber})");
            return true;
        }

        Console.WriteLine($"[Внимание] Клиент {customer.GetFullName()} уже есть в базе!");
        return false;
    }

    // 3. Создание заказа
    public Order CreateOrder(Customer customer, string description, decimal price)
    {
        Order order = new Order(customer, description, price);

        // Подписываем компанию на событие изменения статуса заказа
        order.OnStatusChanged += HandleOrderStatusChanged;

        Orders.Add(order);
        _ordersById.Add(order.OrderId, order);

        Console.WriteLine($"Создан заказ №{order.OrderId}: '{description}' для клиента {customer.GetFullName()}. Сумма: {price} руб.");
        return order;
    }

    // 4. Назначение сотрудника на заказ
    public void AssignEmployeeToOrder(int orderId, Employee employee)
    {
        if (_ordersById.TryGetValue(orderId, out Order? order))
        {
            order.AssignEmployee(employee);
            NotificationService(employee.GetFullName(), $"Вам назначен заказ №{orderId}: {order.Description}");
        }
        else
        {
            Console.WriteLine($"Ошибка: Заказ №{orderId} не найден.");
        }
    }

    // 5. Изменение статуса заказа
    public void ChangeOrderStatus(int orderId, OrderStatus newStatus)
    {
        if (_ordersById.TryGetValue(orderId, out Order? order))
        {
            Console.WriteLine($"Запрос на смену статуса заказа №{orderId} -> {newStatus}:");
            order.SetStatus(newStatus);
        }
        else
        {
            Console.WriteLine($"Ошибка: Заказ №{orderId} не найден.");
        }
    }

    // Обработчик события изменения статуса заказа (Event Handler)
    private void HandleOrderStatusChanged(Order order, OrderStatus oldStatus, OrderStatus newStatus)
    {
        Console.WriteLine($"[СОБЫТИЕ] Заказ №{order.OrderId}: статус изменен с '{oldStatus}' на '{newStatus}'!");

        // Уведомление ответственного сотрудника через делегат
        if (order.AssignedEmployee != null)
        {
            order.AssignedEmployee.Notify($"Статус вашего заказа №{order.OrderId} изменен на '{newStatus}'.");
        }

        // Уведомление заказчика
        order.OrderCustomer.Notify($"Ваш заказ №{order.OrderId} перешел в статус '{newStatus}'.");
    }

    // 6. Получение заказов по статусу (лямбда-выражение LINQ)
    public List<Order> GetOrdersByStatus(OrderStatus status)
    {
        return Orders.Where(o => o.Status == status).ToList();
    }

    // 7. Поиск заказов по фамилии заказчика
    public List<Order> FindOrdersByCustomer(string lastName)
    {
        return Orders.Where(o => o.OrderCustomer.LastName.Equals(lastName, StringComparison.OrdinalIgnoreCase)).ToList();
    }

    // 8. Поиск заказа в словаре по Id
    public bool FindOrderById(int orderId, out Order? order)
    {
        return _ordersById.TryGetValue(orderId, out order);
    }

    public int GetUniqueCustomerCount() => _uniqueCustomers.Count;

    // 9. Отчет по сотрудникам (количество обработанных заказов)
    public void GenerateEmployeeReport()
    {
        Console.WriteLine(new string('-', 70));
        Console.WriteLine($"ОТЧЕТ ПО СОТРУДНИКАМ КОМПАНИИ '{Name.ToUpper()}':");
        Console.WriteLine(new string('-', 70));
        Console.WriteLine($"{"ФИО Сотрудника",-25} | {"Должность",-30} | {"Обработано заказов",18}");
        Console.WriteLine(new string('-', 70));

        foreach (var emp in Employees.OrderByDescending(e => e.OrdersProcessed))
        {
            Console.WriteLine($"{emp.GetFullName(),-25} | {emp.Position,-30} | {emp.OrdersProcessed,18}");
        }
        Console.WriteLine(new string('-', 70));
    }

    // 10. Сводный финансовый отчет
    public void GenerateCompanySummary()
    {
        decimal totalRevenue = Orders.Where(o => o.Status == OrderStatus.Completed).Sum(o => o.Price);
        decimal activeOrdersValue = Orders.Where(o => o.Status == OrderStatus.InProgress || o.Status == OrderStatus.Created).Sum(o => o.Price);

        Console.WriteLine($"Сводка компании: {Name}");
        Console.WriteLine($"Всего сотрудников: {Employees.Count}");
        Console.WriteLine($"Всего заказчиков: {Customers.Count}");
        Console.WriteLine($"Всего заказов в системе: {Orders.Count}");
        Console.WriteLine($"Выручка от завершенных заказов: {totalRevenue:N2} руб.");
        Console.WriteLine($"Потенциальная выручка активных заказов: {activeOrdersValue:N2} руб.");
    }
}
```
