```
using System;
using System.Collections.Generic;
using System.Linq;

Console.OutputEncoding = System.Text.Encoding.UTF8;

// ============================================================================
// 1. ИСХОДНАЯ ДЕМОНСТРАЦИЯ РАБОТЫ С КОМПЬЮТЕРАМИ (СОХРАНЕНА БЕЗ ИЗМЕНЕНИЙ)
// ============================================================================

Computer basicPC = ComputerFactory.CreateBasicPC();
Computer gamingPC = ComputerFactory.CreateGamingPC();

Console.WriteLine("=== БАЗОВЫЙ КОМПЬЮТЕР ===");
basicPC.ShowComponents();
Console.WriteLine("Общая стоимость: " + basicPC.GetTotalPrice() + " руб.\n");

Console.WriteLine("=== ИГРОВОЙ КОМПЬЮТЕР ===");
gamingPC.ShowComponents();
Console.WriteLine("Общая стоимость: " + gamingPC.GetTotalPrice() + " руб.\n");

Console.WriteLine("=== ИНФОРМАЦИЯ О КОМПОНЕНТАХ ===");
CPU cpu1 = new CPU("Intel Core i5", 18000, 6, "Intel", "LGA1700", 3.7, 65);
CPU cpu2 = new CPU("Intel Core i7", 28000, 8, "Intel", "LGA1700", 4.2, 125);

Console.WriteLine(cpu1.GetInfo());
Console.WriteLine(cpu2.GetInfo());
Console.WriteLine();

Console.WriteLine("=== СРАВНЕНИЕ КОМПОНЕНТОВ ===");
Console.WriteLine("Процессоры равны: " + (cpu1 == cpu2)); // перегруженные операторы
Console.WriteLine("Процессоры не равны: " + (cpu1 != cpu2) + "\n");

Console.WriteLine("=== ДОБАВЛЕНИЕ КОМПОНЕНТА ===");
basicPC.AddComponent(cpu1);
basicPC.ShowComponents();
Console.WriteLine("Общая стоимость: " + basicPC.GetTotalPrice() + " руб.\n");

Console.WriteLine("=== УДАЛЕНИЕ КОМПОНЕНТА ===");
basicPC.RemoveComponent(cpu1);
basicPC.ShowComponents();
Console.WriteLine("Общая стоимость: " + basicPC.GetTotalPrice() + " руб.\n");

Console.WriteLine("=== ПОИСК КОМПОНЕНТОВ ПО ТИПУ ===");
gamingPC.ShowComponents<GPU>();
Console.WriteLine();

Console.WriteLine("=== СОЗДАНИЕ КОМПЬЮТЕРА ЧЕРЕЗ ОБОБЩЁННУЮ ФАБРИКУ ===");
Computer cpuComputer = ComputerFactory.CreateComputer(
    new CPU("AMD Ryzen 5", 20000, 6, "AMD", "AM5", 3.8, 65),
    new CPU("Intel Core i5", 18000, 6, "Intel", "LGA1700", 3.7, 65)
);
cpuComputer.ShowComponents();
Console.WriteLine();


// ============================================================================
// 2. ДОПОЛНИТЕЛЬНОЕ ЗАДАНИЕ: ДЕЛЕГАТЫ, ЛЯМБДА, СОБЫТИЯ И КОЛЛЕКЦИИ (ПО ТЕОРИИ)
// ============================================================================

Console.WriteLine(new string('=', 65));
Console.WriteLine("=== ДОП. ЗАДАНИЕ: КОЛЛЕКЦИИ, ДЕЛЕГАТЫ, ЛЯМБДА И СОБЫТИЯ ===");
Console.WriteLine(new string('=', 65));
Console.WriteLine();

// --- А. ДЕЛЕГАТЫ (по теории Task 2) ---
Console.WriteLine(">>> 1. ДЕЛЕГАТЫ:");
DisplayDelegate displayCpu = cpu1.DisplayInfo;
DisplayDelegate displayGpu = new GPU("NVIDIA RTX 4060", 35000, 8, "NVIDIA", 115, 2, "PCIe 4.0").DisplayInfo;

// Вызов методов через делегаты
displayCpu();
displayGpu();
Console.WriteLine();


// --- Б. ЛЯМБДА-ВЫРАЖЕНИЯ (по теории Task 2) ---
Console.WriteLine(">>> 2. ЛЯМБДА-ВЫРАЖЕНИЯ:");
List<Component> allComponents = new List<Component>
{
    cpu1,
    cpu2,
    new CPU("AMD Ryzen 7", 30000, 8, "AMD", "AM5", 4.5, 105),
    new RAM("Kingston Fury", 8000, 32, "Kingston", 3200, "DDR4", 1.35),
    new GPU("NVIDIA RTX 4070", 65000, 12, "NVIDIA", 200, 3, "PCIe 4.0")
};

// Фильтрация через лямбда-выражение (поиск всех процессоров Intel)
var intelProcessors = allComponents.OfType<CPU>().Where(c => c.Manufacturer == "Intel");
Console.WriteLine("Процессоры производителя Intel (отфильтрованы лямбдой):");
foreach (var proc in intelProcessors)
{
    proc.DisplayInfo();
}
Console.WriteLine();


// --- В. СОБЫТИЯ (по теории Task 2) ---
Console.WriteLine(">>> 3. СОБЫТИЯ:");
// Подписываем метод на событие обновления цены компонента
cpu1.ComponentUpdated += OnComponentUpdated;
// Изменяем цену; это автоматически вызовет событие ComponentUpdated
cpu1.UpdatePrice(16500);
Console.WriteLine();


// --- Г. СЛОВАРЬ DICTIONARY<TKey, TValue> (по теории Task 0) ---
Console.WriteLine(">>> 4. КОЛЛЕКЦИЯ DICTIONARY<string, Component>:");
Dictionary<string, Component> componentDictionary = new Dictionary<string, Component>
{
    { "CPU-INTEL-01", cpu1 },
    { "CPU-INTEL-02", cpu2 },
    { "GPU-NV-01", new GPU("RTX 4070", 65000, 12, "NVIDIA", 200, 3, "PCIe 4.0") }
};

// Поиск и вывод компонента по ключу через TryGetValue
if (componentDictionary.TryGetValue("CPU-INTEL-01", out Component foundComponent))
{
    Console.WriteLine("Найден компонент по ключу 'CPU-INTEL-01':");
    foundComponent.DisplayInfo();
}
else
{
    Console.WriteLine("Компонент не найден.");
}
Console.WriteLine();


// --- Д. МНОЖЕСТВО HASHSET<T> (по теории Task 1) ---
Console.WriteLine(">>> 5. КОЛЛЕКЦИЯ HASHSET<Component> (УНИКАЛЬНЫЕ ЭЛЕМЕНТЫ):");
HashSet<Component> componentSet = new HashSet<Component>();

Component compA = new CPU("Intel Core i5", 18000, 6, "Intel", "LGA1700", 3.7, 65);
Component compB = new CPU("AMD Ryzen 7", 30000, 8, "AMD", "AM5", 4.5, 105);
componentSet.Add(compA);
componentSet.Add(compB);

// Попытка добавить дубликат (благодаря переопределенным Equals и GetHashCode дубликат отвергается)
Component duplicateComp = new CPU("Intel Core i5", 18000, 6, "Intel", "LGA1700", 3.7, 65);
bool isAdded = componentSet.Add(duplicateComp);

Console.WriteLine($"Был ли добавлен дублирующий процессор? {isAdded} (False — дубликат не добавлен)");
Console.WriteLine($"Всего уникальных компонентов в множестве: {componentSet.Count}");


// Обработчик события (Task 2)
static void OnComponentUpdated(string message)
{
    Console.WriteLine(message);
}


// ============================================================
// ОПРЕДЕЛЕНИЕ ДЕЛЕГАТА (Task 2)
// ============================================================
public delegate void DisplayDelegate();


// ============================================================
// ИНТЕРФЕЙС И КЛАССЫ С ДОБАВЛЕННЫМИ АТРИБУТАМИ, МЕТОДАМИ И СОБЫТИЯМИ
// ============================================================

// Интерфейс устанавливаемого компонента
public interface IInstallable
{
    void Install();
}

// Абстрактный базовый класс Component
public abstract class Component : IInstallable
{
    // Исходные атрибуты
    public string Name { get; set; }
    public decimal Price { get; set; }

    // --- НОВЫЕ АТРИБУТЫ (3 шт.) ---
    public string Manufacturer { get; set; }      // Производитель
    public int WarrantyMonths { get; set; }       // Срок гарантии (мес.)
    public string Article { get; set; }           // Артикул / серийный номер

    // --- ДЕЛЕГАТ И СОБЫТИЕ (Task 2) ---
    public delegate void ComponentHandler(string message);
    public event ComponentHandler? ComponentUpdated;

    public Component(string name, decimal price, string manufacturer = "Не указан", int warrantyMonths = 12)
    {
        Name = name;
        Price = price;
        Manufacturer = manufacturer;
        WarrantyMonths = warrantyMonths;
        Article = Guid.NewGuid().ToString().Substring(0, 8).ToUpper();
    }

    public virtual string GetInfo()
    {
        return $"Название: {Name}, Производитель: {Manufacturer}, Цена: {Price} руб., Гарантия: {WarrantyMonths} мес.";
    }

    public override string ToString() => GetInfo();

    public abstract void Install();

    // --- НОВЫЕ МЕТОДЫ (3 шт.) ---

    // 1. Метод вывода информации (по теории Task 0, 1, 2)
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"[{GetType().Name}] {Name} (Производитель: {Manufacturer}), Цена: {Price} руб.");
    }

    // 2. Метод обновления цены с вызовом события (Task 2)
    public void UpdatePrice(decimal newPrice)
    {
        decimal oldPrice = Price;
        Price = newPrice;
        // Генерация события
        ComponentUpdated?.Invoke($"[Событие] Цена компонента '{Name}' обновлена: {oldPrice} руб. -> {newPrice} руб.");
    }

    // 3. Метод применения скидки
    public void ApplyDiscount(decimal percent)
    {
        if (percent > 0 && percent <= 100)
        {
            decimal discount = Price * (percent / 100);
            UpdatePrice(Price - discount);
        }
    }

    // --- ПЕРЕОПРЕДЕЛЕНИЕ EQUALS И GETHASHCODE ДЛЯ HASHSET (Task 1) ---
    public override bool Equals(object? obj)
    {
        if (obj is Component other)
        {
            return Name == other.Name && Manufacturer == other.Manufacturer && Price == other.Price;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Manufacturer, Price);
    }
}


// Процессор
public class CPU : Component
{
    public int Cores { get; set; }

    // --- НОВЫЕ АТРИБУТЫ (3 шт.) ---
    public string Socket { get; set; }            // Сокет (LGA1700, AM5)
    public double BaseFrequencyGHz { get; set; }  // Базовая частота (ГГц)
    public int TDP { get; set; }                  // Тепловыделение (Вт)

    public CPU(string name, decimal price, int cores,
               string manufacturer = "Intel", string socket = "LGA1700",
               double frequency = 3.5, int tdp = 65)
        : base(name, price, manufacturer)
    {
        Cores = cores;
        Socket = socket;
        BaseFrequencyGHz = frequency;
        TDP = tdp;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", количество ядер: {Cores}, Сокет: {Socket}, Частота: {BaseFrequencyGHz} ГГц, TDP: {TDP}W";
    }

    public override void Install()
    {
        Console.WriteLine("Установка процессора: " + Name);
    }

    // --- НОВЫЕ МЕТОДЫ (3 шт.) ---
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"   -> Ядер: {Cores}, Сокет: {Socket}, Частота: {BaseFrequencyGHz} ГГц");
    }

    public void Overclock(double additionalGHz)
    {
        BaseFrequencyGHz += additionalGHz;
        Console.WriteLine($"Процессор {Name} разогнан до {BaseFrequencyGHz:F1} ГГц.");
    }

    public bool IsCompatibleWith(string targetSocket)
    {
        return Socket.Equals(targetSocket, StringComparison.OrdinalIgnoreCase);
    }

    public static bool operator ==(CPU? a, CPU? b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price && a.Cores == b.Cores;
    }

    public static bool operator !=(CPU? a, CPU? b) => !(a == b);

    public override bool Equals(object? obj) => base.Equals(obj) && obj is CPU c && Cores == c.Cores;

    public override int GetHashCode() => HashCode.Combine(base.GetHashCode(), Cores);
}


// Оперативная память
public class RAM : Component
{
    public int Capacity { get; set; }

    // --- НОВЫЕ АТРИБУТЫ (3 шт.) ---
    public int FrequencyMHz { get; set; }         // Частота (МГц)
    public string MemoryType { get; set; }        // Тип памяти (DDR4, DDR5)
    public double Voltage { get; set; }           // Напряжение (В)

    public RAM(string name, decimal price, int capacity,
               string manufacturer = "Kingston", int frequency = 3200,
               string memoryType = "DDR4", double voltage = 1.35)
        : base(name, price, manufacturer)
    {
        Capacity = capacity;
        FrequencyMHz = frequency;
        MemoryType = memoryType;
        Voltage = voltage;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", объём памяти: {Capacity} ГБ, Тип: {MemoryType}, Частота: {FrequencyMHz} МГц";
    }

    public override void Install()
    {
        Console.WriteLine("Установка оперативной памяти: " + Name);
    }

    // --- НОВЫЕ МЕТОДЫ (3 шт.) ---
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"   -> Объём: {Capacity} ГБ {MemoryType}, Частота: {FrequencyMHz} МГц");
    }

    public void EnableXMP()
    {
        FrequencyMHz += 400;
        Console.WriteLine($"Профиль XMP активирован для {Name}. Новая частота: {FrequencyMHz} МГц.");
    }

    public void IncreaseVoltage(double delta)
    {
        Voltage += delta;
        Console.WriteLine($"Напряжение памяти {Name} увеличено до {Voltage} В.");
    }

    public static bool operator ==(RAM? a, RAM? b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price && a.Capacity == b.Capacity;
    }

    public static bool operator !=(RAM? a, RAM? b) => !(a == b);

    public override bool Equals(object? obj) => base.Equals(obj) && obj is RAM r && Capacity == r.Capacity;

    public override int GetHashCode() => HashCode.Combine(base.GetHashCode(), Capacity);
}


// Видеокарта
public class GPU : Component
{
    public int Memory { get; set; }

    // --- НОВЫЕ АТРИБУТЫ (3 шт.) ---
    public int PowerConsumptionWatt { get; set; } // Энергопотребление (Вт)
    public int FanCount { get; set; }             // Количество вентиляторов
    public string InterfaceType { get; set; }     // Интерфейс (PCIe 4.0)

    public GPU(string name, decimal price, int memory,
               string manufacturer = "NVIDIA", int powerWatt = 200,
               int fanCount = 3, string interfaceType = "PCIe 4.0")
        : base(name, price, manufacturer)
    {
        Memory = memory;
        PowerConsumptionWatt = powerWatt;
        FanCount = fanCount;
        InterfaceType = interfaceType;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", видеопамять: {Memory} ГБ, Потребление: {PowerConsumptionWatt}W, Охлаждение: {FanCount} кулера";
    }

    public override void Install()
    {
        Console.WriteLine("Установка видеокарты: " + Name);
    }

    // --- НОВЫЕ МЕТОДЫ (3 шт.) ---
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"   -> Видеопамять: {Memory} ГБ GDDR6, Энергопотребление: {PowerConsumptionWatt}W");
    }

    public void SetFanSpeed(int percent)
    {
        Console.WriteLine($"Скорость вентиляторов ({FanCount} шт.) видеокарты {Name} установлена на {percent}%.");
    }

    public bool NeedsAdditionalPower() => PowerConsumptionWatt > 75;

    public static bool operator ==(GPU? a, GPU? b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price && a.Memory == b.Memory;
    }

    public static bool operator !=(GPU? a, GPU? b) => !(a == b);

    public override bool Equals(object? obj) => base.Equals(obj) && obj is GPU g && Memory == g.Memory;

    public override int GetHashCode() => HashCode.Combine(base.GetHashCode(), Memory);
}


// Жёсткий диск
public class HDD : Component
{
    public int Capacity { get; set; }

    // --- НОВЫЕ АТРИБУТЫ (3 шт.) ---
    public int SpindleSpeedRPM { get; set; }      // Скорость вращения (RPM)
    public int CacheSizeMB { get; set; }          // Объем кэша (МБ)
    public string FormFactor { get; set; }        // Форм-фактор (3.5", 2.5")

    public HDD(string name, decimal price, int capacity,
               string manufacturer = "Seagate", int rpm = 7200,
               int cacheMB = 256, string formFactor = "3.5\"")
        : base(name, price, manufacturer)
    {
        Capacity = capacity;
        SpindleSpeedRPM = rpm;
        CacheSizeMB = cacheMB;
        FormFactor = formFactor;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", объём диска: {Capacity} ГБ, Скорость: {SpindleSpeedRPM} RPM, Кэш: {CacheSizeMB} МБ";
    }

    public override void Install()
    {
        Console.WriteLine("Установка жёсткого диска: " + Name);
    }

    // --- НОВЫЕ МЕТОДЫ (3 шт.) ---
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"   -> Объём: {Capacity} ГБ, Скорость шпинделя: {SpindleSpeedRPM} RPM, Форм-фактор: {FormFactor}");
    }

    public void RunDefrag()
    {
        Console.WriteLine($"Дефрагментация диска {Name} ({Capacity} ГБ) успешно завершена.");
    }

    public void CheckHealth()
    {
        Console.WriteLine($"SMART-диагностика диска {Name}: Состояние отличное (100%).");
    }

    public static bool operator ==(HDD? a, HDD? b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price && a.Capacity == b.Capacity;
    }

    public static bool operator !=(HDD? a, HDD? b) => !(a == b);

    public override bool Equals(object? obj) => base.Equals(obj) && obj is HDD h && Capacity == h.Capacity;

    public override int GetHashCode() => HashCode.Combine(base.GetHashCode(), Capacity);
}


// ============================================================
// КЛАСС КОМПЬЮТЕРА И ФАБРИКА
// ============================================================

public class Computer
{
    private List<Component> components = new List<Component>();

    public void AddComponent<T>(T component)
        where T : Component, IInstallable
    {
        if (component == null)
        {
            Console.WriteLine("Нельзя добавить пустой компонент.");
            return;
        }

        component.Install();
        components.Add(component);
        Console.WriteLine("Компонент добавлен: " + component.Name);
    }

    public void AddComponentFromFactory(Component component)
    {
        if (component == null) return;
        components.Add(component);
    }

    public void RemoveComponent<T>(T component)
        where T : Component
    {
        if (components.Remove(component))
            Console.WriteLine("Компонент удалён: " + component.Name);
        else
            Console.WriteLine("Компонент не найден.");
    }

    public decimal GetTotalPrice() => components.Sum(component => component.Price);

    public void ShowComponents()
    {
        if (components.Count == 0)
        {
            Console.WriteLine("Компьютер не содержит компонентов.");
            return;
        }

        foreach (Component component in components)
        {
            Console.WriteLine(component.GetInfo());
        }
    }

    public void ShowComponents<T>() where T : Component
    {
        bool found = false;
        foreach (Component component in components)
        {
            if (component is T)
            {
                Console.WriteLine(component.GetInfo());
                found = true;
            }
        }
        if (!found)
            Console.WriteLine("Компоненты данного типа не найдены.");
    }
}

public static class ComputerFactory
{
    public static Computer CreateComputer<T>(params T[] components)
        where T : Component, IInstallable
    {
        Computer computer = new Computer();
        foreach (T component in components)
            computer.AddComponent(component);
        return computer;
    }

    public static Computer CreateBasicPC()
    {
        Computer computer = new Computer();
        computer.AddComponentFromFactory(new CPU("Intel Core i3", 12000, 4, "Intel", "LGA1700", 3.3, 58));
        computer.AddComponentFromFactory(new RAM("Kingston", 4000, 8, "Kingston", 2666, "DDR4", 1.2));
        computer.AddComponentFromFactory(new HDD("Seagate", 5000, 1000, "Seagate", 7200, 64, "3.5\""));
        return computer;
    }

    public static Computer CreateGamingPC()
    {
        Computer computer = new Computer();
        computer.AddComponentFromFactory(new CPU("AMD Ryzen 7", 30000, 8, "AMD", "AM5", 4.5, 105));
        computer.AddComponentFromFactory(new RAM("Kingston Fury", 8000, 32, "Kingston", 3600, "DDR4", 1.35));
        computer.AddComponentFromFactory(new GPU("NVIDIA RTX 4070", 65000, 12, "NVIDIA", 200, 3, "PCIe 4.0"));
        computer.AddComponentFromFactory(new HDD("WD Black", 7000, 2000, "Western Digital", 7200, 256, "3.5\""));
        return computer;
    }
}
```
