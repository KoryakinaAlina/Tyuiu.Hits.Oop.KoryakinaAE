```
using System;
using System.Collections.Generic;
using System.Linq;

Console.OutputEncoding = System.Text.Encoding.UTF8;

// ============================================================================
// 1. ДЕМОНСТРАЦИЯ ИСХОДНОЙ РАБОТЫ С КОМПЬЮТЕРАМИ (СОХРАНЕНА БЕЗ ИЗМЕНЕНИЙ)
// ============================================================================

Computer basicPC = ComputerFactory.CreateBasicPC();
Computer gamingPC = ComputerFactory.CreateGamingPC();

Console.WriteLine("=== БАЗОВЫЙ КОМПЬЮТЕР ===");
basicPC.ShowComponents();
Console.WriteLine("Общая стоимость: " + basicPC.GetTotalPrice() + " руб.");

Console.WriteLine();

Console.WriteLine("=== ИГРОВОЙ КОМПЬЮТЕР ===");
gamingPC.ShowComponents();
Console.WriteLine("Общая стоимость: " + gamingPC.GetTotalPrice() + " руб.");

Console.WriteLine();

Console.WriteLine("=== ИНФОРМАЦИЯ О КОМПОНЕНТАХ ===");
CPU cpu1 = new CPU("Intel Core i5", 18000, 6);
CPU cpu2 = new CPU("Intel Core i7", 28000, 8);

Console.WriteLine(cpu1.GetInfo());
Console.WriteLine(cpu2.GetInfo());

Console.WriteLine();

Console.WriteLine("=== СРАВНЕНИЕ КОМПОНЕНТОВ ===");
Console.WriteLine("Процессоры равны: " + (cpu1 == cpu2)); // перегруженные операторы
Console.WriteLine("Процессоры не равны: " + (cpu1 != cpu2));

Console.WriteLine();

Console.WriteLine("=== ДОБАВЛЕНИЕ КОМПОНЕНТА ===");
basicPC.AddComponent(cpu1);
basicPC.ShowComponents();
Console.WriteLine("Общая стоимость: " + basicPC.GetTotalPrice() + " руб.");

Console.WriteLine();

Console.WriteLine("=== УДАЛЕНИЕ КОМПОНЕНТА ===");
basicPC.RemoveComponent(cpu1);
basicPC.ShowComponents();
Console.WriteLine("Общая стоимость: " + basicPC.GetTotalPrice() + " руб.");

Console.WriteLine();

Console.WriteLine("=== ПОИСК КОМПОНЕНТОВ ПО ТИПУ ===");
gamingPC.ShowComponents<GPU>();

Console.WriteLine();

Console.WriteLine("=== СОЗДАНИЕ КОМПЬЮТЕРА ЧЕРЕЗ ОБОБЩЁННУЮ ФАБРИКУ ===");
Computer cpuComputer = ComputerFactory.CreateComputer(
    new CPU("AMD Ryzen 5", 20000, 6),
    new CPU("Intel Core i5", 18000, 6)
);
cpuComputer.ShowComponents();

Console.WriteLine();
Console.WriteLine(new string('=', 65));
Console.WriteLine("=== СПРИНТ РЕВЬЮ 4: ПРЕДПРИЯТИЕ, ПОСТАВЩИКИ И СОТРУДНИКИ ===");
Console.WriteLine(new string('=', 65));
Console.WriteLine();


// ============================================================================
// 2. ДЕМОНСТРАЦИЯ СПРИНТ РЕВЬЮ 4 (ШАГ 3 ЗАДАНИЯ)
// ============================================================================

// 1. Создаем объект класса Enterprise
IEnterprise enterprise = new Enterprise("ИТ-Холдинг 'ТехноСфера'");

// 2. Добавляем поставщиков (согласно заданию)
enterprise.AddSupplier(new Supplier("Логистический центр", "Логистика", 100000));
enterprise.AddSupplier(new Supplier("Поставщик офисной мебели", "Офисное оборудование", 50000));

// 3. Добавляем сотрудников (согласно заданию)
enterprise.AddEmployee(new Employee("Иван Иванов", "Менеджер", 70000));
enterprise.AddEmployee(new Employee("Анна Петрова", "Разработчик", 95000));

// 4. Отображаем список поставщиков (в точном формате примера)
Console.WriteLine("Список поставщиков:");
int supplierIndex = 1;
foreach (ISupplier supplier in enterprise.GetSuppliers())
{
    Console.WriteLine($"{supplierIndex++}. {supplier.GetName()} – Стоимость услуг: {supplier.GetCost()}");
}

// Отображаем список сотрудников (в точном формате примера)
Console.WriteLine("Список сотрудников:");
int employeeIndex = 1;
foreach (IEmployee employee in enterprise.GetEmployees())
{
    Console.WriteLine($"{employeeIndex++}. {employee.GetFullName()} – Должность: {employee.Position} – Зарплата: {employee.GetSalary()}");
}

// 5. Вызываем метод подсчёта общих расходов и выводим результат
Console.WriteLine($"Общие расходы: {enterprise.CalculateTotalExpenses()}");

Console.WriteLine();
Console.WriteLine("--- Демонстрация расширяемости (SOLID: Open-Closed Principle) ---");
// Добавление новых специализированных поставщиков и сотрудников без изменения кода Enterprise:
enterprise.AddSupplier(new ComponentSupplier("Intel Distribution", 450000, "Микропроцессоры"));
enterprise.AddEmployee(new LeadEngineer("Сергей Смирнов", "Архитектор компьютерных систем", 150000, "Senior"));

Console.WriteLine($"Новые общие расходы с учетом расширений: {enterprise.CalculateTotalExpenses()} руб.");


// ============================================================================
// ИСХОДНЫЕ КЛАССЫ И ИНТЕРФЕЙСЫ КОМПЬЮТЕРОВ (БЕЗ ИЗМЕНЕНИЙ)
// ============================================================================

public interface IInstallable
{
    void Install();
}

public abstract class Component : IInstallable
{
    public string Name { get; set; }
    public decimal Price { get; set; }

    public Component(string name, decimal price)
    {
        Name = name;
        Price = price;
    }

    public virtual string GetInfo()
    {
        return "Название: " + Name + ", цена: " + Price + " руб.";
    }

    public override string ToString()
    {
        return GetInfo();
    }

    public abstract void Install();
}

public class CPU : Component
{
    public int Cores { get; set; }

    public CPU(string name, decimal price, int cores)
        : base(name, price)
    {
        Cores = cores;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + ", количество ядер: " + Cores;
    }

    public override void Install()
    {
        Console.WriteLine("Установка процессора: " + Name);
    }

    public static bool operator ==(CPU a, CPU b)
    {
        if (ReferenceEquals(a, b))
            return true;

        if (a is null || b is null)
            return false;

        return a.Name == b.Name && a.Price == b.Price && a.Cores == b.Cores;
    }

    public static bool operator !=(CPU a, CPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is CPU other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price, Cores);
    }
}

public class RAM : Component
{
    public int Capacity { get; set; }

    public RAM(string name, decimal price, int capacity)
        : base(name, price)
    {
        Capacity = capacity;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + ", объём памяти: " + Capacity + " ГБ";
    }

    public override void Install()
    {
        Console.WriteLine("Установка оперативной памяти: " + Name);
    }

    public static bool operator ==(RAM a, RAM b)
    {
        if (ReferenceEquals(a, b))
            return true;

        if (a is null || b is null)
            return false;

        return a.Name == b.Name && a.Price == b.Price && a.Capacity == b.Capacity;
    }

    public static bool operator !=(RAM a, RAM b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is RAM other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price, Capacity);
    }
}

public class GPU : Component
{
    public int Memory { get; set; }

    public GPU(string name, decimal price, int memory)
        : base(name, price)
    {
        Memory = memory;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + ", видеопамять: " + Memory + " ГБ";
    }

    public override void Install()
    {
        Console.WriteLine("Установка видеокарты: " + Name);
    }

    public static bool operator ==(GPU a, GPU b)
    {
        if (ReferenceEquals(a, b))
            return true;

        if (a is null || b is null)
            return false;

        return a.Name == b.Name && a.Price == b.Price && a.Memory == b.Memory;
    }

    public static bool operator !=(GPU a, GPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is GPU other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price, Memory);
    }
}

public class HDD : Component
{
    public int Capacity { get; set; }

    public HDD(string name, decimal price, int capacity)
        : base(name, price)
    {
        Capacity = capacity;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + ", объём диска: " + Capacity + " ГБ";
    }

    public override void Install()
    {
        Console.WriteLine("Установка жёсткого диска: " + Name);
    }

    public static bool operator ==(HDD a, HDD b)
    {
        if (ReferenceEquals(a, b))
            return true;

        if (a is null || b is null)
            return false;

        return a.Name == b.Name && a.Price == b.Price && a.Capacity == b.Capacity;
    }

    public static bool operator !=(HDD a, HDD b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is HDD other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price, Capacity);
    }
}

public class Computer
{
    private List<Component> components = new List<Component>();

    public void AddComponent<T>(T component)
        where T : Component, IInstallable
    {
        if (component == null)
        {
            Console.WriteLine("Нельзя добавить пустой компонент.");
            return;
        }

        component.Install();
        components.Add(component);
        Console.WriteLine("Компонент добавлен: " + component.Name);
    }

    public void AddComponentFromFactory(Component component)
    {
        if (component == null)
            return;

        component.Install();
        components.Add(component);
    }

    public void RemoveComponent<T>(T component)
        where T : Component
    {
        if (components.Remove(component))
            Console.WriteLine("Компонент удалён: " + component.Name);
        else
            Console.WriteLine("Компонент не найден.");
    }

    public decimal GetTotalPrice()
    {
        return components.Sum(component => component.Price);
    }

    public void ShowComponents()
    {
        if (components.Count == 0)
        {
            Console.WriteLine("Компьютер не содержит компонентов.");
            return;
        }

        foreach (Component component in components)
        {
            Console.WriteLine(component.GetInfo());
        }
    }

    public void ShowComponents<T>() where T : Component
    {
        bool found = false;

        foreach (Component component in components)
        {
            if (component is T)
            {
                Console.WriteLine(component.GetInfo());
                found = true;
            }
        }

        if (!found)
            Console.WriteLine("Компоненты данного типа не найдены.");
    }
}

public static class ComputerFactory
{
    public static Computer CreateComputer<T>(params T[] components)
        where T : Component, IInstallable
    {
        Computer computer = new Computer();

        foreach (T component in components)
        {
            computer.AddComponent(component);
        }

        return computer;
    }

    public static Computer CreateBasicPC()
    {
        Computer computer = new Computer();

        computer.AddComponentFromFactory(new CPU("Intel Core i3", 12000, 4));
        computer.AddComponentFromFactory(new RAM("Kingston", 4000, 8));
        computer.AddComponentFromFactory(new HDD("Seagate", 5000, 1000));

        return computer;
    }

    public static Computer CreateGamingPC()
    {
        Computer computer = new Computer();

        computer.AddComponentFromFactory(new CPU("AMD Ryzen 7", 30000, 8));
        computer.AddComponentFromFactory(new RAM("Kingston Fury", 8000, 32));
        computer.AddComponentFromFactory(new GPU("NVIDIA RTX 4070", 65000, 12));
        computer.AddComponentFromFactory(new HDD("WD Black", 7000, 2000));

        return computer;
    }
}


// ============================================================================
// НОВЫЕ СУЩНОСТИ СПРИНТ РЕВЬЮ 4: ШАГ 1 (ИНТЕРФЕЙСЫ)
// ============================================================================

// 1. Интерфейс IEnterprise для взаимодействия с поставщиками и сотрудниками
public interface IEnterprise
{
    void AddSupplier(ISupplier supplier);
    void AddEmployee(IEmployee employee);
    List<ISupplier> GetSuppliers();
    List<IEmployee> GetEmployees();
    decimal CalculateTotalExpenses();
}

// 2. Интерфейс ISupplier для поставщиков
public interface ISupplier
{
    string GetName();
    decimal GetCost();
    string SupplierType { get; set; }
}

// 3. Интерфейс IEmployee для сотрудников
public interface IEmployee
{
    string GetFullName();
    decimal GetSalary();
    string Position { get; set; }
}


// ============================================================================
// НОВЫЕ СУЩНОСТИ СПРИНТ РЕВЬЮ 4: ШАГ 2 (РЕАЛИЗАЦИЯ КЛАССОВ)
// ============================================================================

// 1. Класс Enterprise, имплементирующий IEnterprise
public class Enterprise : IEnterprise
{
    public string EnterpriseName { get; set; }

    // Хранение коллекций через интерфейсы (управление зависимостями)
    private readonly List<ISupplier> _suppliers = new List<ISupplier>();
    private readonly List<IEmployee> _employees = new List<IEmployee>();

    public Enterprise(string enterpriseName = "Главное предприятие")
    {
        EnterpriseName = enterpriseName;
    }

    public void AddSupplier(ISupplier supplier)
    {
        if (supplier == null)
            throw new ArgumentNullException(nameof(supplier), "Поставщик не может быть null.");

        _suppliers.Add(supplier);
    }

    public void AddEmployee(IEmployee employee)
    {
        if (employee == null)
            throw new ArgumentNullException(nameof(employee), "Сотрудник не может быть null.");

        _employees.Add(employee);
    }

    public List<ISupplier> GetSuppliers()
    {
        return _suppliers;
    }

    public List<IEmployee> GetEmployees()
    {
        return _employees;
    }

    public decimal CalculateTotalExpenses()
    {
        decimal totalSuppliers = _suppliers.Sum(s => s.GetCost());
        decimal totalEmployees = _employees.Sum(e => e.GetSalary());
        return totalSuppliers + totalEmployees;
    }
}

// 2. Класс Supplier, имплементирующий ISupplier
public class Supplier : ISupplier
{
    public string Name { get; set; }
    public decimal Cost { get; set; }
    public string SupplierType { get; set; }

    public Supplier(string name, string supplierType, decimal cost)
    {
        Name = name;
        SupplierType = supplierType;
        Cost = cost;
    }

    public string GetName() => Name;

    public decimal GetCost() => Cost;
}

// 3. Класс Employee, имплементирующий IEmployee
public class Employee : IEmployee
{
    public string FullName { get; set; }
    public string Position { get; set; }
    public decimal Salary { get; set; }

    public Employee(string fullName, string position, decimal salary)
    {
        FullName = fullName;
        Position = position;
        Salary = salary;
    }

    public string GetFullName() => FullName;

    public decimal GetSalary() => Salary;
}


// ============================================================================
// ПРИМЕРЫ РАСШИРЯЕМОСТИ (SOLID / Open-Closed Principle)
// ============================================================================

// Поставщик компьютерных комплектующих
public class ComponentSupplier : Supplier
{
    public string Category { get; set; }

    public ComponentSupplier(string name, decimal cost, string category)
        : base(name, "ИТ-оборудование и комплектующие", cost)
    {
        Category = category;
    }
}

// Специализированный инженер предприятия
public class LeadEngineer : Employee
{
    public string QualificationLevel { get; set; }

    public LeadEngineer(string fullName, string position, decimal salary, string qualificationLevel)
        : base(fullName, position, salary)
    {
        QualificationLevel = qualificationLevel;
    }
}
```
